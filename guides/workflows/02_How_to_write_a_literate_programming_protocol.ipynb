# How to write a literate programming protocol (simple rules)

This notebook defines a minimal standard for writing **literate protocols** in this repository.

**Scope:** Only covers **Workflows** and the **Unit operations** that make up workflows.


## ID conventions

### Workflow ID convention
Workflow identifiers follow the format:

`W-O-TT-XX`

- **W** – Workflow
- **O** – Opentrons (platform)
- **TT** – Assay / test category
- **XX** – Sequential number within the category

**Assay category codes** (extend as needed):
- **PS** – Phosphate solubilisation
- **NF** – Nitrogen fixation

Examples:
- `W-O-PS-01`
- `W-O-NF-02`

### Unit operation ID convention
Unit operations follow the format:

`U-<type>-XX`

- **U** – Unit operation
- **type** – Category of unit operation
  - **O** – Opentrons robot operation (liquid handling)
  - **C** – Clariostar operation (e.g., plate reader)
- **XX** – Sequential number within that unit operation category

Examples:
- `U-O-03` (Unit operation-Opentron-Number)
- `U-C-02` (Unit operation-Clariostar-Number)


## 1) One notebook = one workflow
- Each protocol notebook implements **exactly one** workflow ID (e.g. `W-O-PS-02`).
- Recommended filename starts with the workflow ID, e.g. `W-O-PS-02_phosphate_liquid_assay.ipynb`.


## 2) Always start with a controlled header (SOP block)
The first markdown cell must contain:
- **Doc No** (e.g. `SOP W-O-PS-01`)
- **Version**
- **Effective date**
- **Owner**
- **Approved by**
- **Workflow ID**
- **Workflow name**
- **Unit operations used (in order)**
- **Output** (1 sentence)

This turns the notebook into a controlled protocol document rather than just notes.


## 3) Unit operations are the table of contents
Early in the notebook, include a section:

## Steps (unit operations)
1. **U-...** Unit operation name
2. **U-...** Unit operation name
3. **U-...** Unit operation name

This should let a reader understand the protocol in ~10 seconds.


## 4) Parameters live in one place
- Have **exactly one** "Parameters" cell near the top.
- Use clear names and include units in the name when relevant (e.g. `MEDIA_VOL_UL`).
- Avoid "magic numbers" elsewhere in the notebook.


## 5) Every unit operation uses the same mini-format
Each unit operation section starts with:

### U-XX-YY Unit operation name
- **Goal:** (1 line)
- **Inputs:** (plates, reagents, files)
- **Outputs:** (what is produced / where saved)
- **Acceptance criteria:** (what must be true to proceed)
- **Record:** (what to log in run notes)
- **Deviation handling:** (what to do if criteria fail)
- **Notes:** (tips, contamination, mixing, timing)

Keep it brief and operational.


## 6) Light separation of story vs execution
For each unit operation:
- 1–3 markdown bullets describing **intent**
- followed by the **code cell(s)** that implement it


## 7) Always end with artifacts
At the bottom of the notebook, clearly show:
- Where exported artifacts are written (e.g. generated robot script path)
- A short run summary (e.g. key parameters used)


## 8) Validation gates are mandatory before release
Each workflow notebook should define explicit checks before it is considered runnable:
- Input schema checks (required columns, value ranges, duplicates)
- Output integrity checks (row counts, unique wells, barcode/primer resolution)
- Script validation (`python -m py_compile` for unit operation scripts)
- Simulation validation (`opentrons_simulate` commands documented and run)
- Acceptance statement: what constitutes a pass/fail run


## 9) End with a controlled change log
At notebook end, include a short table:
- Date
- Version
- Author
- Summary of change

This makes protocol evolution auditable.
